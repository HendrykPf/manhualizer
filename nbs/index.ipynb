# manhualizer

> A tool that converts stories into manhua style comics

## Developer Setup

### Prerequisites

- Python 3.10+
- [uv](https://docs.astral.sh/uv/) for environment and package management

### Clone and install

```sh
git clone https://github.com/HendrykPf/manhualizer.git
cd manhualizer

# Create virtual environment and install dependencies
uv venv
uv sync
```

### Development workflow

All source code lives in notebooks under `nbs/`. Do not edit files in `manhualizer/` directly — they are auto-generated.

```sh
# Export notebooks to Python modules
uv run nbdev-export

# Run tests
uv run nbdev-test

# Clean, test, export, and build docs (run before pushing)
uv run nbdev-prepare
```

### Install in editable mode (optional)

```sh
uv pip install -e .
```

## Usage

### Installation

Install latest from the [GitHub repository](https://github.com/HendrykPf/manhualizer):

pip install git+https://github.com/HendrykPf/manhualizer.git
